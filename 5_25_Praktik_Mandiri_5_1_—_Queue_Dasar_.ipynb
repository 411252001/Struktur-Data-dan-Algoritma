```python
class Queue:
  def __init__(self):
    self.items = []

  def is_empty(self):
    return len(self.items) ==0

  def enqueue(self, data):
    self.items.append(data)

  def dequeue(self):
    if self.is_empty():
      return None
    return self.items.pop(0)

  def peek(self):
    if self.is_empty():
      return None
    return self.items[0]

  def size(self):
    return len(self.items)


urutan = Queue()
print(urutan.is_empty())
urutan.enqueue(10)
urutan.enqueue(20)
urutan.enqueue(30)
urutan.enqueue(40)
urutan.enqueue(50)

print("Tampilkan Size :", urutan.size())
print("Tampilkan Front :", urutan.peek())
print("Dequeue Pertama :", urutan.dequeue())
print("Dequeue Kedua :", urutan.dequeue())
print("Tambahkan 60:", urutan.enqueue(60))
print("Tambahkan 70:", urutan.enqueue(70))
urutan.items

print("Kosongkan queue dengan loop")
while not urutan.is_empty():
  print("Mengkosongkan", urutan.dequeue())

print(urutan.is_empty())

#Uji dequeue dan peek ketika queue kosong
print("Dequeue :", urutan.dequeue())
print("Peek :", urutan.peek())
```
