```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return None

    def is_empty(self):
        return len(self.items) == 0

def is_balanced(seimbang):
    stack = Stack()
    pairs = {
        '(': ')',
        '{': '}',
        '[': ']'
    }
    for char in seimbang:
        if char in '({[':
            stack.push(char)
        elif char in ')}]':
            if stack.is_empty():
                return False
            top = stack.pop()
            if pairs[top] != char:
                return False
    return stack.is_empty()

print('(a+b) =', is_balanced('(a+b)'))
print('([{}])', is_balanced('([{}])'))
print('([)]', is_balanced('([)]'))
print('((a+b)', is_balanced('((a+b)'))
print('{a+[b*c]}', is_balanced('{a+[b*c]}'))
```

### Dry Run Table untuk String `([{}])`

Berikut adalah proses penelusuran (dry run) langkah demi langkah saat memeriksa keseimbangan tanda kurung pada string `([{}])`:

| Langkah | Karakter | Jenis Karakter | Aksi | Isi Stack (Bottom -> Top) |
| :---: | :---: | :---: | :---: | :---: |
| **Inisialisasi** | - | - | Inisialisasi Stack kosong | `[]` |
| **1** | `(` | Pembuka | Push `(` ke dalam stack | `['(']` |
| **2** | `[` | Pembuka | Push `[` ke dalam stack | `['(', '[']` |
| **3** | `{` | Pembuka | Push `{` ke dalam stack | `['(', '[', '{']` |
| **4** | `}` | Penutup | Pop `{` cocok dengan `}` | `['(', '[']` |
| **5** | `]` | Penutup | Pop `[` cocok dengan `]` | `['(']` |
| **6** | `)` | Penutup | Pop `(` cocok dengan `)` | `[]` |

**Hasil Akhir:** Stack kosong di akhir proses (`is_empty() == True`), sehingga string dinyatakan **Seimbang (True)**.
