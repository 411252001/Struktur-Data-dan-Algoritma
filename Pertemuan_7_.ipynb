```python
# 7.5.2
def bubble_sort(data):
    n = len(data)

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]

    return data

nilai = [5, 1, 4, 2, 8]
print(bubble_sort(nilai))
```

```python
# 7.6. Bubble Sort dengan Early Stop
def bubble_sort_optimized(data):
    n = len(data)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
                swapped = True

        if not swapped:
            break

    return data
```

```python
# 7.7. Menghitung Perbandingan dan Swap pada Bubble Sort
def bubble_sort_stats(data):
    comparisons = 0
    swaps = 0
    n = len(data)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            comparisons += 1
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
                swaps += 1

    return data, comparisons, swaps

hasil, c, s = bubble_sort_stats([5, 1, 4, 2, 8])
print(hasil)
print('Perbandingan:', c)
print('Swap:', s)
```

```python
# 7.3 Pemilihan nilai minimum pada Selection Sort
def selection_sort(data):
    n = len(data)
    for i in range(n - 1):
        min_index = i

        for j in range(i + 1, n):
            if data[j] < data[min_index]:
                min_index = j

        data[i], data[min_index] = data[min_index], data[i]

    return data

nilai = [29, 10, 14, 37, 13]
print(selection_sort(nilai))
```

```python
# 7.4 Penyisipan key pada Insertion Sort
def insertion_sort(data):
  for i in range(1, len(data)):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

nilai = [5, 8, 12, 4, 9]
print(insertion_sort(nilai))
```

```python
# 7.13. Fungsi Merge: Inti dari Merge Sort
def merge(left, right):
  result = []
  i = 0
  j = 0
  while i < len(left) and j < len(right):
    if left[i] <= right[j]:
      result.append(left[i])
      i += 1
    else:
      result.append(right[j])
      j += 1
  result.extend(left[i:])
  result.extend(right[j:])

  return result

print(merge([3, 9, 27], [10, 38, 43]))
```

```python
# 7.14. Implementasi Merge Sort
def merge_sort(data):
    if len(data) <= 1:
        return data

    mid = len(data) // 2
    left = merge_sort(data[:mid])
    right = merge_sort(data[mid:])

    return merge(left, right)

nilai = [38, 27, 43, 3, 9, 82, 10]
print(merge_sort(nilai))
```

```python
# 7.15. Melihat Jejak Rekursi Merge Sort
def merge_sort_trace(data, depth=0):
  indent = '  ' * depth
  print(indent + 'masuk:', data)

  if len(data) <= 1:
    print(indent + 'return:', data)
    return data

  mid = len(data) // 2
  left = merge_sort_trace(data[:mid], depth + 1)
  right = merge_sort_trace(data[mid:], depth + 1)
  result = merge(left, right)

  print(indent + 'merge :', result)
  return result

merge_sort_trace([8, 3, 5, 1])
```
