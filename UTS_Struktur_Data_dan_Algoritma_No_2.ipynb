```python
# ==========================================
# 1. STRUKTUR DATA (NODE, QUEUE, STACK)
# ==========================================

class Node:
    def __init__(self, data):
        self.data = data # Simpan ID mahasiswa
        self.next = None # Pointer ke elemen berikutnya, default kosong

class LinkedQueue:
    def __init__(self):
        self.front = None # Pointer penunjuk elemen terdepan
        self.rear = None  # Pointer penunjuk elemen terbelakang
        self.count = 0    # Penghitung jumlah antrean

    def is_empty(self):
        return self.count == 0 # Bernilai True jika antrean kosong

    def size(self):
        return self.count # Mengembalikan total antrean saat ini

    def enqueue(self, data):
        new_node = Node(data) # Cetak node baru
        if self.is_empty():
            self.front = self.rear = new_node # Jika kosong, depan & belakang = node baru
        else:
            self.rear.next = new_node # Sambungkan elemen terakhir ke node baru
            self.rear = new_node # Pindahkan status 'belakang' ke node baru
        self.count += 1 # Tambah jumlah antrean

    def enqueue_front(self, data):
        new_node = Node(data) # Cetak node baru untuk Undo
        if self.is_empty():
            self.front = self.rear = new_node # Jika kosong, depan & belakang = node baru
        else:
            new_node.next = self.front # Sambungkan node baru ke elemen terdepan saat ini
            self.front = new_node # Pindahkan status 'depan' ke node baru
        self.count += 1 # Tambah jumlah antrean

    def dequeue(self):
        if self.is_empty():
            print("  [Aman] Error ditangani: Queue kosong.") # Handle edge case
            return None
        data = self.front.data # Simpan data terdepan sebelum dihapus
        self.front = self.front.next # Geser pointer depan ke elemen kedua
        self.count -= 1 # Kurangi jumlah antrean
        if self.front is None:
            self.rear = None # Jika antrean habis, reset pointer belakang ke None
        return data # Kembalikan data mahasiswa yang dilayani

    def peek(self):
        return self.front.data if not self.is_empty() else None # Intip data terdepan

    def display(self):
        elems, curr = [], self.front # Siapkan list tampungan dan mulai dari depan
        while curr:
            elems.append(curr.data) # Masukkan string data ke list
            curr = curr.next # Pindah ke node selanjutnya
        print("  Queue : " + (" -> ".join(elems) if elems else "KOSONG")) # Cetak visual

class Stack:
    def __init__(self):
        self.items = [] # Gunakan list Python sebagai tumpukan riwayat

    def is_empty(self): return len(self.items) == 0 # True jika list kosong
    def size(self): return len(self.items) # Kembalikan panjang list riwayat
    def push(self, data): self.items.append(data) # Masukkan data ke ujung list (Top)

    def pop(self):
        if self.is_empty():
            print("  [Aman] Error ditangani: Stack riwayat kosong.") # Handle edge case
            return None
        return self.items.pop() # Cabut dan ambil data dari ujung list

    def peek(self): return self.items[-1] if not self.is_empty() else None # Intip data terakhir
    def display(self): print("  Stack :", self.items if self.items else "KOSONG") # Cetak visual


# ==========================================
# 2. PERSIAPAN & SIMULASI 12 LANGKAH
# ==========================================

d = 1 # Digit terakhir NIM Anda
A, B, C, D, E, F = [f'MHS-10{i}' for i in range(1, 7)] # Buat ID MHS-101 s/d MHS-106 otomatis

queue = LinkedQueue() # Bikin objek antrean aktif
stack = Stack() # Bikin objek riwayat layanan

def status(langkah, op):
    print(f"\nLangkah {langkah}. {op}") # Cetak nomor langkah dan operasi
    queue.display() # Tampilkan isi Queue
    stack.display() # Tampilkan isi Stack

print("=== MULAI SIMULASI ===")
queue.enqueue(A); status(1, f"DATANG {A}") # Langkah 1: A masuk belakang
queue.enqueue(B); status(2, f"DATANG {B}") # Langkah 2: B masuk belakang
queue.enqueue(C); status(3, f"DATANG {C}") # Langkah 3: C masuk belakang

stack.push(queue.dequeue()); status(4, "LAYANI") # Langkah 4: Layani 1 (A), simpan ke riwayat
queue.enqueue(D); status(5, f"DATANG {D}") # Langkah 5: D masuk belakang
stack.push(queue.dequeue()); status(6, "LAYANI") # Langkah 6: Layani 1 (B), simpan ke riwayat
queue.enqueue(E); status(7, f"DATANG {E}") # Langkah 7: E masuk belakang

queue.enqueue_front(stack.pop()); status(8, "UNDO") # Langkah 8: Batal layani (B), kembalikan ke depan
stack.push(queue.dequeue()); status(9, "LAYANI") # Langkah 9: Layani ulang (B)
stack.push(queue.dequeue()); status(10, "LAYANI") # Langkah 10: Layani 1 lagi (C)

queue.enqueue_front(stack.pop()); status(11, "UNDO") # Langkah 11: Batal layani (C), kembalikan ke depan
queue.enqueue(F); status(12, f"DATANG {F}") # Langkah 12: F masuk belakang


# ==========================================
# 3. KONDISI AKHIR & UJI EDGE CASE
# ==========================================

print("\n=== HASIL EKSPLISIT PASCA SIMULASI ===")
print(f"Front (Depan)  : {queue.front.data if queue.front else 'Kosong'}") # Siapa yang ada di front
print(f"Rear (Belakang): {queue.rear.data if queue.rear else 'Kosong'}") # Siapa yang ada di rear
print(f"Sisa antrean Queue: {queue.size()} | Sisa riwayat Stack: {stack.size()}") # Total sisa elemen

print("\n=== UJI EDGE CASE ===")
test_q, test_s = LinkedQueue(), Stack() # Bikin objek kosong baru
test_q.dequeue() # Paksa dequeue pada queue kosong
test_s.pop() # Paksa pop pada stack kosong
print("(Pengujian berhasil: program tidak crash saat data kosong)")
```
