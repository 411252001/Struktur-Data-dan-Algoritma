```python
class Kotak:
  def __init__(self, data):
    self.data = data
    self.next = None

class SinglyLinkedList:
  def __init__(self):
    self.head = None
  def insert_beginning(self, data):
    new_node = Kotak(data)
    new_node.next = self.head
    self.head = new_node
  def append(self, data):
    new_node = Kotak(data)
    if self.head is None:
      self.head = new_node
      return
    current = self.head
    while current.next is not None:
      current = current.next
    current.next = new_node

  def display(self):
    current = self.head
    while current is not None:
      print(current.data, end=' -> ')
      current = current.next
    print('None')

linked = SinglyLinkedList()
linked.append(100)
linked.append(200)
linked.append(300)
linked.display()
```

```python
class Kotak:
  def __init__(self, data):
    self.data = data
    self.next = None

class SinglyLinkedList:
  def __init__(self):
    self.head = None
  def insert_beginning(self, data):
    new_node = Kotak(data)
    new_node.next = self.head
    self.head = new_node
  def append(self, data):
    new_node = Kotak(data)
    if self.head is None:
      self.head = new_node
      return
    current = self.head
    while current.next is not None:
      current = current.next
    current.next = new_node

  def display(self):
    current = self.head
    while current is not None:
      print(current.data, end=' -> ')
      current = current.next
    print('None')

linked = SinglyLinkedList()
linked.append(100)
linked.append(200)
linked.append(300)
linked.append(400)

# langsung ke node terakhir
if linked.head and linked.head.next:
    last_node = linked.head
    while last_node.next is not None:
        last_node = last_node.next
    linked.head.next = last_node

linked.display()
```
