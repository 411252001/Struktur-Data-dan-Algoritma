```python
d = 1
data = [
    ['TI001', 'Alya', 78],
    ['TI002', 'Bima', 80 + d],
    ['TI003', 'Citra', 67],
    ['TI004', 'Danu', 92],
    ['TI005', 'Eka', 74],
    ['TI006', 'Fajar', 88],
    ['TI007', 'Gina', 81],
    ['TI008', 'Hadi', 65 + d],
    ['TI009', 'Intan', 91],
    ['TI010', 'Joko', 76],
    ['TI011', 'Karin', 84],
    ['TI012', 'Lukman', 70 + d]
]

# ==========================================
# POIN 1: LINEAR SEARCH
# ==========================================
def linear_search_nim(data, target): # mendefinisikan untuk fungsi linear search
    comparisons = 0 # counter untuk menghitung berapa kali melakukan perbandingan, dimulai dari 0 karena awal
    for i in range(len(data)): # mengecek data di index mulai dari awal ketika mulai melakukan looping
        comparisons += 1 # ketikan melakukan pengecekan karena perbandingan maka akan bertambah
        if data[i][0] == target: # disinilah dilakukan perbandingan apakah data pada index sama dengan data target
            return i, comparisons # jika ketemu maka berhenti melakukan perbandingan
    return -1, comparisons # jika tidak ketemu maka -1 atau data "not found"

# ==========================================
# POIN 2: PENENTUAN TARGET & UJI LINEAR SEARCH
# ==========================================
target_index = d % len(data)  # disini dengan melakukan modulo dari d = 1 dengan jumlah total mahasiswa. 1 % 12 = 1(ambil data dari index ini)
target_nim = data[target_index][0]  # disini mengambil item pertama dari index 1 yaitu nim TI002

# Uji target yang ada dan yang tidak ada
idx_ls_found, comp_ls_found = linear_search_nim(data, target_nim) # jika target ada
idx_ls_notfound, comp_ls_notfound = linear_search_nim(data, 'TI999') # jika tidak ada

# ==========================================
# POIN 3: MERGE SORT (REKURSIF)
# ==========================================
def merge(left, right): # mendefinisikan fungsi merge sorting
    result = [] # awal dari sorting
    i = j = 0 # kedua data dimulai dari indeks 0
    # Gabungkan dua subarray yang sudah terurut
    while i < len(left) and j < len(right): # mulai perbandingan antara sublist kiri dan kanan
        if left[i][0] <= right[j][0]: # jika i lebih kecil/sama dengan j
            result.append(left[i]) # maka ambil data dari i dan tambahkan/simpan ke result
            i += 1 # lalu lanjutkan, geser dari kiri kesebelahnya
        else: # jika sebaliknya, dimana i lebih besar dari j
            result.append(right[j]) # maka ambil data dari j dan taruh di result
            j += 1 # geser penunjuk indeks tumpukan kanan maju 1 langkah ke kartu berikutnya.

    # Masukkan sisa elemen jika salah satu subarray sudah habis
    result.extend(left[i:]) # sisa dari i
    result.extend(right[j:]) # sisa dari j
    return result

def merge_sort_nim(data):
    # Basis kasus rekursi
    if len(data) <= 1:
        return data

    mid = len(data) // 2
    left = merge_sort_nim(data[:mid])
    right = merge_sort_nim(data[mid:])

    return merge(left, right)

# ==========================================
# POIN 4: TAMPILKAN DATA SEBELUM & SESUDAH
# ==========================================
print("=== Poin 4: Data Sebelum dan Sesudah Pengurutan ===")
print("Data Sebelum:")
for baris in data:
    print(baris)

# Gunakan salinan data agar data awal tidak berubah
data_urut = merge_sort_nim(data.copy())

print("\nData Sesudah (Merge Sort):")
for baris in data_urut:
    print(baris)

# ==========================================
# POIN 5: BINARY SEARCH
# ==========================================
def binary_search_nim(data_urut, target):
    comparisons = 0
    low = 0
    high = len(data_urut) - 1

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if data_urut[mid][0] == target:
            return mid, comparisons
        elif data_urut[mid][0] < target:
            low = mid + 1  # Cari di bagian kanan
        else:
            high = mid - 1 # Cari di bagian kiri

    return -1, comparisons # Mengembalikan -1 jika tidak ditemukan

# ==========================================
# POIN 6: UJI BINARY SEARCH
# ==========================================
idx_bs_found, comp_bs_found = binary_search_nim(data_urut, target_nim)
idx_bs_notfound, comp_bs_notfound = binary_search_nim(data_urut, 'TI999')

# ==========================================
# POIN 8: RINGKASAN HASIL
# ==========================================
print("\n=== Poin 8: Ringkasan Hasil Pencarian ===")
print(f"Target Pencarian: {target_nim}")
print(f"Indeks Hasil (Linear Search): {idx_ls_found}")
print(f"Indeks Hasil (Binary Search): {idx_bs_found}")
print(f"Jumlah Perbandingan Linear Search (Target {target_nim}): {comp_ls_found}")
print(f"Jumlah Perbandingan Binary Search (Target {target_nim}): {comp_bs_found}")
print(f"\nUji Kasus Tidak Ditemukan ('TI999'):")
print(f"Linear Search - Indeks: {idx_ls_notfound}, Perbandingan: {comp_ls_notfound}")
print(f"Binary Search - Indeks: {idx_bs_notfound}, Perbandingan: {comp_bs_notfound}")
```

### 9. Analisis Algoritma dan Kompleksitas

**Kompleksitas Waktu:**
*   **Linear Search:** $O(N)$. Algoritma ini memeriksa setiap elemen satu per satu. Pada skenario terburuk (data di akhir list atau tidak ditemukan), ia harus melakukan iterasi sebanyak $N$ kali.
*   **Binary Search:** $O(\log N)$. Algoritma secara sistematis mereduksi ruang pencarian menjadi setengahnya pada setiap iterasi.
*   **Merge Sort:** $O(N \log N)$. Algoritma selalu membagi data menjadi dua bagian hingga berukuran 1 (membutuhkan $\log N$ langkah reduksi), kemudian melakukan proses *merge* yang iterasi totalnya proporsional terhadap panjang array $N$ di setiap tingkat rekursi.

**Kebutuhan Ruang Tambahan (Space Complexity) Merge Sort:**
Merge Sort bukanlah algoritma *in-place*. Dalam implementasi Python menggunakan *list slicing* dan array `result` tambahan, kompleksitas ruangnya adalah $O(N)$. Array baru harus dialokasikan di memori saat mengeksekusi operasi penggabungan (`merge(left, right)`).

**Alasan Binary Search Memerlukan Data Terurut:**
Logika eliminasi Binary Search bergantung pada sifat *monotonicity* (keterurutan). Dengan membandingkan target terhadap nilai tengah (*midpoint*), algoritma dapat mengambil kesimpulan deduktif tentang di belahan mana target tersebut mungkin berada. Jika data acak, nilai tengah tidak memberikan informasi yang dapat diandalkan mengenai posisi elemen-elemen lainnya, sehingga eliminasi separuh wilayah pencarian menjadi mustahil secara matematis.

**Kapan Biaya Sorting Layak Dibayar?**
Biaya komputasi (waktu) untuk melakukan pengurutan $O(N \log N)$ bernilai ekonomis apabila pencarian (kueri) dilakukan secara berulang/masif pada himpunan data yang sama. Jika kita hanya mencari data satu kali, melakukan Linear Search $O(N)$ jauh lebih efisien dibandingkan menyortirnya lalu mencari $O(N \log N + \log N)$. Namun, jika data tersebut dicari sebanyak $K$ kali, asimtotik waktu $O(N \log N + K \log N)$ akan mendominasi dan mengalahkan *brute-force* $O(K \times N)$ seiring bertambah besarnya nilai pengali $K$.
