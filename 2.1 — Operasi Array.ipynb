```python
Nilai = [11,21,31,41,51,61,71,81,91,101]

print(len(Nilai))
print('Nilai Awal =', Nilai) # Kondisi Awal
print('nilai Pertama =', Nilai[1]) # Indeks Pertama
print('Nilai Kelima =', Nilai[5]) # Indeks Kelima
print('Nilai Terakhir =', Nilai[len(Nilai)-1]) # Indeks Terakhir
Nilai[3] = 7 # diubah
print('Nilai pada indeks ke 3 berubah =', Nilai)
Nilai.append(122) # Append
print('Nilai setelah ditambah elemen dengan append =',Nilai)
Nilai.insert(2, 67) # Insert
print('Nilai setelah menambah elemen pada indeks 2 menggunakan insert =', Nilai)
Dihapus = Nilai.pop(7) # POP
print('Nilai Dihapus =', Dihapus)
print('Indeks setelah dihapus dengan pop =',Nilai)
Nilai.remove(67) # Remove
print('Nilai setelah elemen 67 dihapus dengan remove =', Nilai)
for i, x in enumerate(Nilai): # Traversal
  print('Indeks', i, 'nilai', x)
```
