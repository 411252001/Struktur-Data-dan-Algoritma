```python
class Stack:
  def __init__(self):
    self.items = []

  def empty(self):
    return len(self.items) == 0

  def push(self, item):
    self.items.append(item)

  def pop(self):
    if self.empty():
      return None
    return self.items.pop()

  def peek(self):
    if self.empty():
      return None
    return self.items[-1]

  def size(self):
    return len(self.items)

  def clear(self):
    self.items = []


stack = Stack()
print(stack.empty())
stack.push(10)
stack.push(20)
stack.push(30)
stack.push(40)
print('Top:',stack.peek())
print('Size:', stack.size())
pop_1 = stack.pop()
pop_2 = stack.pop()
print('Pop:', [pop_1, pop_2])
stack.push(50)
stack.push(60)
print(stack.items)
stack.clear()
print(stack.items)
print('Pop:', stack.pop()) # Jika melakukan pop pada stack yang kosong maka akan mencetak None, karena sudah di definisikan pada def pop untuke mencetak None jika stack kosong
```
